```python
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv("/content/HR-Employee-Attrition-Cleaned..csv")
print("Total_Records:",len(df))
```

```python
print(df.head())
```

```python
print(df.columns.tolist())
```

```python
data=pd.crosstab(df["Department"], df["Attrition"])
data.plot(kind="bar", figsize=(6, 5))
plt.title("Employee Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)
plt.legend(title="Attrition")
plt.tight_layout()
plt.show()
```

```python
counts=df["Attrition"].value_counts()
plt.pie(counts,labels=counts.index, autopct="%1.1f%%")
plt.title("Employee Attrition Rate")
plt.show()
```

```python
data=pd.crosstab(df["OverTime"],df["Attrition"])
data.plot(kind="bar")
plt.title("Attrition by OverTime")
plt.xlabel("OverTime")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)
plt.legend(title="Attrition")
plt.show()
```

```python
data=pd.crosstab(df["JobSatisfaction"],df["Attrition"])
data.plot(kind="bar")
plt.title("Attrition by Job Satisfaction")
plt.xlabel("Job Satisfaction")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)
plt.legend(title="Attrition")
plt.show()
```

```python
data=pd.crosstab(df["YearsAtCompany"],df["Attrition"])
data.plot(kind="line",marker="o")
plt.title("Attrition by Years at Company")
plt.xlabel("Years at Company")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)
plt.legend(title="Attrition")
plt.show()
```

```python
plt.hist(df["Age"],bins=10,edgecolor="black")
plt.title("Age Distribution of Employees")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.show()
```
